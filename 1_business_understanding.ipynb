Project Name:-

Superstore Sales Analysis

problem Statement:-

A retail company wants to analyze its historical sales data to understand customer purchasing behavior, identify profitable products, optimize regional performance, and improve overall business profitability.

Business Objective:-

Analyze sales data to discover patterns, identify high-performing products and customers, evaluate regional performance, and provide actionable business recommendations.

Business Questions

Which category generates the most sales?

Which category generates the most profit?

Which customer spends the most?

Which region performs best?

Which products should be discontinued?

Which months have the highest sales?

What is the effect of discounts on profit?
